#### Minimum detectable exoplanet radius

Say we consider a transiting planet to be conclusively detected when it has a signal-to-noise ratio (SNR) of 3 or greater (the higher the SNR, the more confident we are that the signal is real). This means that the transit depth $\delta$ must be $\geq 3\times$ the noise floor of our detector (i.e., the uncertainty in each data point). 

Here, let's adopt a detector noise floor of 20 ppm, meaning $\delta_{min} = 60$ ppm or $\delta_{min} = 60\times10^{-6}$ for a positive exoplanet detection. You will need to research the state-of-the-art instruments to find your own value for the noise (and cite your source!), but it should be similar-ish to the value I'm using.

The figure below illustrates this concept for $\delta = 0\times$noise (i.e., just the noise of our detector), $\delta = 1\times$noise (no detection), $\delta = 3\times$noise (minimum criteria for conclusive detection), $\delta = 5\times$noise (strong detection).

```python
SNR = [0.0, 1.0, 3.0, 5.0] #signal-to-noise ratio
num_measurements = 200
noise = 20*(10**(-6))

yticks = np.linspace(-500, 500, 41)

fig, ax = plt.subplots(1,len(SNR), figsize = (len(SNR)*5,5))

for i in np.arange(0, len(SNR))[::-1]:
    normalized_flux_without_planet = sp.norm.rvs(1.0, noise, size = num_measurements)
    lc_truth, orb_phase = lc_model(SNR[i]*noise, npts = num_measurements)
    normalized_flux_with_planet = lc_truth + sp.norm.rvs(0.0, noise, num_measurements)

    delta_flux_with_planet = (normalized_flux_with_planet-1)/(10**(-6)) #in ppm
    delta_flux_without_planet = (normalized_flux_without_planet-1)/(10**(-6))
    ax[i].plot(orb_phase, delta_flux_with_planet , 'ko', markersize = 5, label = 'w/ exoplanet')
    ax[i].plot(orb_phase, delta_flux_without_planet , 'ro', alpha = 0.3, markersize = 5, label = 'w/o exoplanet')

    if i == len(SNR)-1:
        yrange = [0.99*min(delta_flux_with_planet), 1.01*max(delta_flux_with_planet)]
        
    ax[i].set_title(f"$\delta = {int(SNR[i])}$" + r'$\times$' + f'{noise/(10**(-6)):.0f} ppm' + '\n' + f'SNR = {SNR[i]:.0f}', fontsize = 16)
    ax[i].set_ylim(yrange)
    ax[i].patch.set_edgecolor('black')  
    ax[i].patch.set_linewidth(2)  
    ax[i].tick_params(which = 'major', axis = 'x', direction = 'in', length = 10, top = True,  right = True, width = 2, labelsize = 16)
    ax[i].tick_params(which = 'minor', direction = 'in',length = 7,  top = True,  right = True, width = 2)
    if i != 0:
        ax[i].tick_params(which = 'major', axis = 'y', direction = 'in', length = 10, top = True,  right = True, width = 2, labelsize = 0)
    else:
        ax[i].tick_params(which = 'major', axis = 'y', direction = 'in', length = 10, top = True,  right = True, width = 2, labelsize = 16)
        ax[i].set_ylabel(r'$\Delta$Flux (ppm)', fontsize = 20)
        ax[i].legend(fontsize = 16)

        
    ax[i].set_xlabel(r'Orbital Phase', fontsize = 20)
    ax[i].xaxis.set_major_locator(MultipleLocator(0.001))
    ax[i].xaxis.set_minor_locator(MultipleLocator(0.00025))
    ax[i].yaxis.set_major_locator(MultipleLocator(25))
    ax[i].yaxis.set_minor_locator(MultipleLocator(5))

plt.tight_layout()
```

OK, great, we have a minimum $\delta$ value that is required for us to consider an exoplanet "detected". But what we *REALLY* want to know is the minimum detectable $R_p$. Solving the $\delta$ equation for planet radius gives

$$\LARGE R_p = \sqrt{\delta}R_\star$$

subbing in $\delta_{min}$ gives the minimum detectable planet radius

$$\LARGE R_{p,min} = \sqrt{\delta_{min}}R_\star = \sqrt{60\times10^{-6}} R_\star \sim 0.008 R_\star$$

What should we use for $R_\star$? Again, this is up to you and your group. You will need to make a logical assumption about $R_\star$ (you can also assume a range of $R_\star$ if you are so inclined). For the sake of this demo, let's adopt $R_\star = 1 R_\odot$. Why? If we look at the mean $R_\star$ for planets detected via the transit method, we get an average value that is effectively equal to 1$R_\odot$:

```python
ind = np.where((data['discoverymethod'] == 'Transit') & (np.isnan(data['st_rad']) == False))

print('Mean R_star for planets discovered via transit photometry: ', np.mean(data['st_rad'].iloc[ind]))
```

If we want to be less sensitive to large stars in our analysis, what statistic would have been a better choice?

Anyway, now that we have a characteristic values for $R_\star$, we can write

$$\LARGE R_{p,min} \sim 0.008\times R_\odot = 0.008\times 109 R_\oplus \sim 0.9R_\oplus$$

#### Maximum exoplanet orbital period / semi-major axis
So we have $R_{p, min}$, but this, of course, exoplanet size is not the only limitation of the transit photometry method. To simply discussion for now, let's consider a single system with 1 star and 1 transiting exoplanet with an orbital period of $P$ days. As we have discussed earlier in the course, the transit of the exoplanet will most likely be very short in duration relative to its $P$. So, if we don't have a good idea of when the exoplanet will transit, if we observe the star for any $t<P$, we are likely to miss the transit and miss our opportunity to measure $R_p$.

Let's generalize this now. We know that the majority of known transiting exoplanets were discovered by the Kepler mission. So, let's use this as our baseline for determining the approximate maximimum $P$ below which we should be able to detect exoplanets reliably.

```python
print('Frac of transiting planets detected by Kepler = ', str(round(len(np.where(data['disc_facility'] == 'Kepler')[0])/len(np.where(data['discoverymethod'] == 'Transit')[0]), 2)))
```

The primary mission of Kepler stared at the same field of view (i.e., the same patch of sky and the same sample of stars) for about 3 years. This means that the planets detected by Kepler would have been limited to orbital periods of $\lesssim 3$ years. Why?

## Part 4: estimating occurence rates / detection biases

Up to this point, we have implicitly assumed that if it is *possible* for us to detect an exoplanet with a given technique, then we *will* detect it. In other words, we are assuming each detection technique has a 100% detection efficiency within the limits you derived for Part 1 above. 

In practice, of course, an instrument or technique can never acheive a 100% detection efficiency. Further, exoplanet detection efficiencies are generally a function of the exoplanet parameters you are trying to measure (e.g., $M_p$, $R_p$, and/or $P$ depending), among other things including the an exoplanet host's characteristics. Additionally, with giant surveys like Kepler, it's not logistically feasible to parse all of the data manually. Instead, we must rely on automated pipelines to flag promising segments of data for potential exoplanet signals. Such pipelines are usually tested by injecting synthetic exoplanetary signals, and asking how many are accurately identified. Spoiler alert, they never recover 100% of the injected signals. Further, data collection is never continuous. An instrument will generally collect a new data point every $\Delta t$ seconds/hours/days/etc. So, if the phenomena you are trying to detect (e.g., a transit) is shorter than your observing cadence, you'll miss it. Anyway... all this is to say that, despite our best efforts, we gon miss a lot of exoplanets that we expect to be able to detect for many, many different reasons.

As such, to accurately determine the relative occurrence rates between the various flavors of exoplanets -- which is incredibly important for understanding planet formation/evolution processes -- it is imperative to account for such biases. For example, if we want to know how many super-Earths there are relative to sub-Neptunes, we need to be able to estimate how many super-Earths/sub-Neptunes we missed for every super-Earth/sub-Neptune we didn't miss.

You can stack or average over multiple transits to get a higher signal to noise given. The signal to noise over multiple transits for Kepler is approximately given by the expression below (taken from [https://iopscience.iop.org/article/10.1088/0067-0049/201/2/15/pdf](https://iopscience.iop.org/article/10.1088/0067-0049/201/2/15/pdf)).     $$\large SNR = \frac{\delta}{\sigma_{CDPP}} \sqrt{\frac{n_{tr} \cdot t_{dur}}{3 hr}}$$
