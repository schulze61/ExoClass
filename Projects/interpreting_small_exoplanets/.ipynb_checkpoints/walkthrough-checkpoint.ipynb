# Project: Interpreting Small Exoplanets

In this project, you will work to identify compositional outliers from a sample of small, potentiall rocky exoplanets assuming that such exoplanets should follow the rock-star relationship. This will require you to combine and expand upon a large number topics we have discussed up to this point: the basic picture of rocky planet formation, mass-radius-composition relationships, host star abundances, the small-planet radius gap, working with the NASA Exoplanet Archive (NEA), and more!

More explicitly, you will be *testing* the following research statement (which we have already debunked thoroughly): Any exoplanet with an orbital period ($P$) and radius ($R_p$) that places it below (in $P-R_p$ space) the small-planet radius gap should be small, gas-free, and water-free with an Fe core mass fractions that mirrors that of its host star, i.e., it should follow the rock-star relationship. 

We can also rewrite this in terms of a testable null hypothesis ($H_0$), e.g., 

$H_0:$ A small exoplanet whose $P-R_p$ lie below the small-planet radius gap will follow the rock-star relationship. 

In this framework, we would specify some statistical significance ($\alpha$) a priori. For example, we could set $\alpha = 0.05$ which means that if a given exoplanet has a $<5\%$ chance of following the rock-star relationship, then we reject $H_0$ in favor of an alternative hypothesis. e.g., the exoplanet is an Fe-enriched super-Mercury or water/steam world. If the exoplanet has a $>5\%$ chance of following the rock-star relationship, we are unable to reject $H_0$, and we would report something along the lines of "We find planet (insert some long catalog name + exoplanet letter here) to consistent with $H_0$ given its current observational uncertainties."

#### Step 1: The first thing we are going to do is load in Fe, Mg, and Si abundances from the Hypatia catalog (https://www.hypatiacatalog.com/hypatia) and convert them to molar Fe/Mg and Si/Mg ratios
    I'm going to do this step for you with a short explanation since this more along the lines of 'Stellar Astronomy' than planetary science, but it is still good to be aware of the basics b/c it does come up in PS.
    
    Stellar abundance values in Hypatia and pretty much everywhere in the literature are usually given as [X/H] where 
    
$$\left[\frac{X}{H}\right] = \log_{10}\left(\frac{N_{X}}{N_{H}}\right)_\star - \log_{10}\left(\frac{N_{X}}{N_{H}}\right)_\odot$$

    
    so the abundance of element X is normalized to that of Hydrogen (usually just assumed to be 10^12) and compared to the Sun. This is a similar concept to Stellar magnitude scales. For iron, this would be
    
$$\left[\frac{Fe}{H}\right] = \log_{10}\left(\frac{N_{Fe}}{N_{H}}\right)_\star - \log_{10}\left(\frac{N_{Fe}}{N_{H}}\right)_\odot$$
    
    
    
    We want to get these into molar ratios of Fe/Mg and Si/Mg to calculate planet composition:
    
$$\Large \frac{Fe}{Mg} = \frac{10^{[Fe/H] \; + \; \log_{10}(N_{Fe}/N_{H})_\odot}}{10^{[Mg/H] \; + \; \log_{10}(N_{Mg}/N_{H})_\odot}}$$

    and
    
 $$\Large \frac{Si}{Mg} = \frac{10^{[Fe/H] \; + \; \log_{10}(N_{Si}/N_{H})_\odot}}{10^{[Mg/H] \; + \; \log_{10}(N_{Mg}/N_{H})_\odot}}$$   
    

In the cell below I'm using values for $\log_{10}(N_{Fe}/N_{H})_\odot$, $\log_{10}(N_{Mg}/N_{H})_\odot$, and $\log_{10}(N_{Si}/N_{H})_\odot$ from Lodders+ 2009 (https://materials.springer.com/lb/docs/sm_lbs_978-3-540-88055-4_34).
