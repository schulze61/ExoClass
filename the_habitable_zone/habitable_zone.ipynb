```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
tfreeze = 275.0
tboil = 375.0
Lsun = 3.846*(10**26)
sb = 5.670374419*(10**(-8))
au2m = 1.496e+11

def amin(L):
    return np.sqrt((L*Lsun)/(16.0*np.pi*sb*(375.0**4)))/au2m
def amax(L):
    return np.sqrt((L*Lsun)/(16.0*np.pi*sb*(275.0**4)))/au2m

def Tp(L, a):
    return 0.5*(((L*Lsun)/(np.pi*sb*((a*au2m)**2)))**0.25)
```

```python
x = np.linspace(0.1, 2.0, 100)

plt.plot(x, Tp(1.0, x), 'r-')
#plt.plot([amin(1.0), amin(1.0)], [0.0, 375.0], 'k--')
#plt.plot([0.1, amin(1.0)], [375.0, 375.0], 'k--')

#plt.plot([amax(1.0), amax(1.0)], [0.0, 275.0], 'k--')
#plt.plot([0.1, amax(1.0)], [275.0, 275.0], 'k--')
plt.xlim(0.0, 2.0)
plt.ylim(0, 800)

plt.ylabel('Eq. Temp (K)', fontsize = 16)
plt.xlabel('a [au]', fontsize = 16)

x = np.linspace(amin(1.0), amax(1.0), 100)
plt.fill_between(x, Tp(1.0, x), color = 'g', alpha = 0.3)

#plt.axvline(1.0)
#plt.axvline(0.72)
#lt.axvline(0.4)
#plt.axvline(1.52)
#plt.axhline(275.0)
#plt.axhline(375.0)


lstar = 0.000522

x = np.linspace(0.001, 2.0, 1000)

plt.plot(x, Tp(lstar, x), 'r-')
#plt.plot([amin(lstar), amin(lstar)], [0.0, 375.0], 'k--')
#plt.plot([0.0, amin(lstar)], [375.0, 375.0], 'k--')

#plt.plot([amax(lstar), amax(lstar)], [0.0, 275.0], 'k--')
#plt.plot([0.0, amax(lstar)], [275.0, 275.0], 'k--')
#plt.xlim(0.0, 0.07)
plt.ylim(0, 800)

plt.ylabel('Eq. Temp (K)', fontsize = 16)
plt.xlabel('a [au]', fontsize = 16)

x = np.linspace(amin(lstar), amax(lstar), 100)
plt.fill_between(x, Tp(lstar, x), color = 'g', alpha = 0.3)
```
