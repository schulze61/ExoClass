# Angular separation

![orbital_setup.png](./figures/orbital_setup.png)

The figure above illustrates the point that a planet and its host orbit their mutual center of mass, meaning each have their own orbital semi-major axes that are not equal, but rather related by 

$$\LARGE M_p \times a_\mathrm{planet} = M_\star \times a_\mathrm{star}$$

**Important**: for the remainder of this notebook, we will use $a$ for the semi-major axis of the planet ($a = a_\mathrm{planet}$), as is usually standard in astronomy. When discussing the semi-major axis of a stars orbit, we will use $a_\star$.

Now imagine the system above is at a distance $d$ away from you, the observer. The *maximum* angular seperation ($\theta$) you will measure between the exoplanet and its host is

$$\LARGE \theta \simeq \frac{a}{d}$$

### Arcminutes and arcseconds
In astronomy, in exoplanetary science, we deal with systems that are incredibly far away from us and therefore have very, very small angular seperations. Therefore, when quoting $\theta$ values, using degrees usually doesn't give us fine enough resolution. Instead, we will usually work in $\mu$as, or microarcseconds. # Direct Imaging

![planetary_society_direct_imaging_graphic.png](./figures/planetary_society_direct_imaging_graphic.png)

The trick with direct imaging is 
1. determining whether an exoplanet will be resolvable from its host with your telescope (Rayleigh criterion)
2. trying to find the wavelength range where the light from the exoplanet relative to its host is maximized (star-planet contrast)
3. determining whether you can detect the maximum contrast with your detector, i.e., the contrast is significantly larger than the noise floor of the detector. 

## The Rayleigh criterion

For a telescope with diameter $D$ observing at some wavelength $\lambda$, the minimum resolvable angular seperation is determined by the [Rayleigh criterion](http://hyperphysics.phy-astr.gsu.edu/hbase/phyopt/Raylei.html).

$$\LARGE \theta_{min} \simeq 1.22 \frac{\lambda}{D}$$

To resolve the planet from the star then the following must be satisfied:

$$\LARGE \frac{a}{d} \geq \theta_{min}$$

Just looking at the equation above, discuss with your classmates what $a$ and $d$ are favored/disfavored by this technique.

## Star-planet contrast

Assuming the Rayleigh criterion is satisfied for whatever exoplanet you are looking at, then whether or not you can actually image the planet depends on how much light it gives off and how much it constrasts the host star. The planet will both emit its own light (blackbody radiation) and reflect some fraction of the host star's light. The contrast between star and planet ($f$) is then the combination of the contrast between the thermal emission of star and planet ($f_\mathrm{thermal}$) and the fraction of star light that the planet reflects ($f_\mathrm{reflected}$), or

$$\LARGE f = f_\mathrm{reflected} + f_\mathrm{thermal}$$

The maximum fraction of the star's light that is reflected by a planet is given via:

$$\LARGE f_\mathrm{reflected} = \frac{\pi R_p^2}{4\pi a^2}A_g$$

where $A_g$ is the planet's geometric albedo, which is a measure of the fraction of incoming light from the host that is reflected by the exoplanet. A planet with a rocky surface will have a lower albedo than a planet with an icy surface. With your classmates, see if you can derive the equation for $f_\mathrm{reflected}$ above. Hint: if you haven't thought about flux in a while, it might be worth a quick refresher.

The star-planet contrast of thermal light is given via: 

$$\LARGE f_\mathrm{thermal} = \left(\frac{R_p}{R_\star}\right)^2 \frac{B_\lambda (T_p)}{B_\lambda (T_\star)} = \left(\frac{R_p}{R_\star}\right)^2 \frac{e^{hc/\lambda k T_\star} - 1}{e^hc/\lambda k T_p - 1}$$

where $T_p$ is the temperature of the planet, $T_\star$ is the temperature of the star, $h$ is Planck's constant, $c$ is the speed of light, $k$ is Boltzmann's constant, and $B_\lambda$ is the Planck function expressed in terms of wavelength. You will often see the Planck function expressed in terms of frequency ($\nu$), or $B_\nu$. 

Here, we will assume that $T_p$ is the equilibrium temperature of the planet, or 

$$\LARGE T_p = T_\star (1-A_b)^{1/4} \sqrt{\frac{R_\star}{2a}}.$$

where $A_b$ is the planet's bond albedo. Check out this [hyperphysics](http://hyperphysics.phy-astr.gsu.edu/hbase/phyopt/albedo.html) page for a brief description of the difference between $A_g$ and $A_b$ if you are curious. Their values are usually similar-ish for a given object.

We will discuss $T_p$ in more detail when we discuss the habitable zone. For now, it is worth noting that the approximation above can be wrong by a good margin. However, unless we have more information to inform the $T_p$ of an exoplanet, it is usually the best we can do when modelling and trying to identify the best targets for direct imaging. How does the $T_p$ for Venus compare with its actual surface temperature?

# Astrometry

##### <span style="color:white">Astrometry</span>


Of the detection methods discussed thus far, astrometry is perhaps simplist mathematically and the most intuitive (to me, at least, but your mileage may vary). However, it remains quite challenging to detect planets via this technique in practice. Like the RV method, detecting planets via astrometry exploits the fact that an unseen planetary companion and its host star orbit their mutual center of mass. If the orbital semi-major axis of the host is large enough relative to the detection limits of our instrument(s), then we can observe the periodic motion of the host star relative to effectively fixed background stars. The astrometric signal $\theta$ of a planet of mass $M_p$ orbiting a star of mass $M_\star$ with a semi-major axis $a$ that is $d$ away from the observer is given by

$$\LARGE\theta = \left(\frac{M_p}{M_\star}\right)\frac{a}{d}$$

Work with your coursemates to derive this expression and discuss what types of planets are favored by this technique. Remember that $a$ is the semi-major axis of the exoplanet.

Similar to direct imaging, there is a minimum $\theta$ that we can resolve. Plugging in some $\theta_{min}$ for our detector and solving for $M_p$ gives the minimum planet mass that we can resolve as a function of $d$, $a$, and $\theta_{min}$. Rearranging the expression above for $M_p$ (since this is the variable we are trying to measure) gives

$$\LARGE M_{p,min} = \theta_{min} M_\star \frac{d}{a}$$

Just looking at this equation, how does the minimum measureable planet mass change with each of the parameters?

The plot below shows a color map of the minimum detectable planet mass around a star with $M_{star}$ as a function of $a$ and $d$ for an assumed $\theta_{min}$. I have set $\theta_{min}=\theta_{min}\sim 400 \mu \mathrm{as}$ (micro arcseconds) to roughly correspond to a significant signal with a "state-of-the-art" astrometric instrument (e.g., [GRAVITY on VLTI](https://www.eso.org/sci/facilities/paranal/instruments/gravity.html) and [GAIA](https://www.esa.int/Science_Exploration/Space_Science/Gaia_overview), with precisions of $\sim$10s-100 $\mu$as). Any planet masses specified in the 'mass_levels' array will show up as black lines on the plot. It is currently set up to show Earth, Neptune, and Jupiter mass planets.
